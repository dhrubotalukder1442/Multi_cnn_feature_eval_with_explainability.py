```python
import os
import numpy as np
import cv2
from PIL import Image
import matplotlib.pyplot as plt

import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import GlobalAveragePooling2D

# Keras model imports & preprocessors
from tensorflow.keras.applications.vgg16 import VGG16, preprocess_input as vgg_pre
from tensorflow.keras.applications.resnet50 import ResNet50, preprocess_input as resnet_pre
from tensorflow.keras.applications.mobilenet_v2 import MobileNetV2, preprocess_input as mob_pre
from tensorflow.keras.applications.inception_resnet_v2 import InceptionResNetV2, preprocess_input as ir_pre
from tensorflow.keras.applications.nasnet import NASNetMobile, preprocess_input as nas_pre
from tensorflow.keras.applications.xception import Xception, preprocess_input as xcp_pre

from sklearn.utils import shuffle
from sklearn.model_selection import StratifiedKFold
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    f1_score, roc_auc_score, matthews_corrcoef, cohen_kappa_score,
    confusion_matrix, accuracy_score
)

# LIME imports
from lime import lime_image
from skimage.segmentation import mark_boundaries

# === Dataset loader (keeps your original approach) ===
def load_images_pil(main_folder_path, target_size=(224, 224)):
    images = []
    labels = []
    for label in ["benign", "malignant"]:
        label_path = os.path.join(main_folder_path, label)
        if not os.path.isdir(label_path):
            continue
        for file in os.listdir(label_path):
            if not file.lower().endswith(('.png', '.jpg', '.jpeg')):
                continue
            img_path = os.path.join(label_path, file)
            try:
                img = Image.open(img_path).convert('RGB')
                img = img.resize(target_size)
                img_array = np.array(img)
                images.append(img_array)
                labels.append(0 if label == "benign" else 1)
            except Exception as e:
                print(f"[WARNING] Skipped {img_path}: {e}")
    return np.array(images), np.array(labels)

# === Helper: find convolutional layers ===
def is_conv_layer(layer):
    # conv-like layers have rank-4 output shapes
    shape = getattr(layer, 'output_shape', None)
    if shape is None:
        return False
    return len(shape) == 4

def get_nth_conv_layer(model, n=3):
    conv_layers = [l for l in model.layers if is_conv_layer(l)]
    if len(conv_layers) == 0:
        raise ValueError("No convolutional layers found")
    idx = min(n-1, len(conv_layers)-1)
    return conv_layers[idx]

def get_last_conv_layer(model):
    conv_layers = [l for l in model.layers if is_conv_layer(l)]
    if len(conv_layers) == 0:
        raise ValueError("No convolutional layers found")
    return conv_layers[-1]

# === Mapping of models & preprocessors ===
MODEL_ZOO = {
    "VGG16":      (VGG16, vgg_pre),
    "ResNet50":   (ResNet50, resnet_pre),
    "MobileNetV2":(MobileNetV2, mob_pre),
    "InceptionResNetV2": (InceptionResNetV2, ir_pre),
    "NASNetMobile": (NASNetMobile, nas_pre),
    "Xception":   (Xception, xcp_pre),
}

# === Feature extraction function (returns features for all images) ===
def extract_features_for_model(model_name, X, batch_size=32):
    ModelClass, preprocess = MODEL_ZOO[model_name]
    print(f"[INFO] Loading base model for features: {model_name}")
    base = ModelClass(weights='imagenet', include_top=False, input_shape=(224,224,3), pooling='avg')
    model = base  # includes global pooling by pooling='avg'
    # Preprocess
    X_copy = X.astype(np.float32)
    X_prep = preprocess(X_copy.copy())
    feats = model.predict(X_prep, batch_size=batch_size, verbose=1)
    return feats

# === Generic feature-map visualization for an intermediate conv layer ===
def visualize_feature_maps_generic(model_name, image, conv_index=3, num_maps=8):
    ModelClass, preprocess = MODEL_ZOO[model_name]
    print(f"[INFO] Visualizing {model_name} internal feature maps (conv index {conv_index})...")
    base = ModelClass(weights='imagenet', include_top=False, input_shape=(224,224,3))
    target_layer = get_nth_conv_layer(base, conv_index)
    fmap_model = Model(inputs=base.input, outputs=target_layer.output)

    img_resized = cv2.resize(image, (224,224)).astype(np.float32)
    inp = np.expand_dims(img_resized, axis=0)
    inp = preprocess(inp)
    fmap = fmap_model.predict(inp)
    fmap = fmap[0]
    # show first num_maps channels
    fig, axs = plt.subplots(2, (num_maps+1)//2, figsize=(12,6))
    axs = axs.ravel()
    for i in range(num_maps):
        if i >= fmap.shape[-1]:
            axs[i].axis('off')
            continue
        axs[i].imshow(fmap[:,:,i], cmap='viridis')
        axs[i].set_title(f'{model_name} fmap {i+1}')
        axs[i].axis('off')
    plt.suptitle(f"{model_name} - conv layer: {target_layer.name}")
    plt.tight_layout()
    plt.show()

# === Grad-CAM generic (uses last conv layer) ===
def compute_gradcam_generic(model_name, image):
    ModelClass, preprocess = MODEL_ZOO[model_name]
    print(f"[INFO] Grad-CAM using pretrained {model_name} (include_top=True) ...")
    # load full model with top for class scores
    full = ModelClass(weights='imagenet', include_top=True, input_shape=(224,224,3))
    last_conv = get_last_conv_layer(full)
    grad_model = Model(inputs=full.input, outputs=[last_conv.output, full.output])

    img_resized = cv2.resize(image, (224,224)).astype(np.float32)
    inp = np.expand_dims(img_resized, axis=0)
    inp = preprocess(inp)

    with tf.GradientTape() as tape:
        conv_outputs, predictions = grad_model(inp)
        # choose top predicted class
        top_index = tf.argmax(predictions[0])
        loss = predictions[0, top_index]

    grads = tape.gradient(loss, conv_outputs)
    pooled_grads = tf.reduce_mean(grads, axis=(0,1,2))
    conv_outputs = conv_outputs[0].numpy()
    pooled_grads = pooled_grads.numpy()
    heatmap = np.zeros(conv_outputs.shape[:2], dtype=np.float32)
    for i in range(conv_outputs.shape[-1]):
        heatmap += pooled_grads[i] * conv_outputs[:,:,i]
    heatmap = np.maximum(heatmap, 0)
    if np.max(heatmap) != 0:
        heatmap /= np.max(heatmap)

    heatmap = cv2.resize(heatmap, (224,224))
    heatmap = np.uint8(255 * heatmap)
    heatmap_colored = cv2.applyColorMap(heatmap, cv2.COLORMAP_JET)
    overlay = cv2.addWeighted(image.astype(np.uint8), 0.6, heatmap_colored, 0.4, 0)
    return overlay

# === LIME generic explanation (uses include_top True ImageNet model) ===
def explain_with_lime_generic(model_name, image, num_samples=1000):
    ModelClass, preprocess = MODEL_ZOO[model_name]
    print(f"[INFO] LIME explanation using pretrained {model_name} (include_top=True) ...")
    full = ModelClass(weights='imagenet', include_top=True, input_shape=(224,224,3))
    explainer = lime_image.LimeImageExplainer()

    def predict_fn(imgs):
        imgs = imgs.astype(np.float32)
        imgs_prep = preprocess(imgs.copy())
        preds = full.predict(imgs_prep)
        return preds

    explanation = explainer.explain_instance(image.astype(np.uint8), predict_fn, top_labels=1, hide_color=0, num_samples=num_samples)
    top_label = explanation.top_labels[0]
    temp, mask = explanation.get_image_and_mask(label=top_label, positive_only=True, hide_rest=False)
    img_bound = mark_boundaries(temp/255.0, mask)
    plt.imshow(img_bound)
    plt.title(f"LIME ({model_name}) top label {top_label}")
    plt.axis('off')
    plt.show()

# === Metrics utilities ===
import math
from sklearn.metrics import roc_auc_score

def compute_fold_metrics(y_true, y_pred, y_proba):
    f1 = f1_score(y_true, y_pred)
    try:
        auc = roc_auc_score(y_true, y_proba[:,1])
    except Exception:
        auc = float('nan')
    mcc = matthews_corrcoef(y_true, y_pred)
    kappa = cohen_kappa_score(y_true, y_pred)
    return f1, auc, mcc, kappa

def mean_and_normal_ci(values, alpha=0.05):
    arr = np.array(values, dtype=np.float64)
    n = len(arr)
    mean = np.nanmean(arr)
    std = np.nanstd(arr, ddof=1) if n > 1 else 0.0
    z = 1.96  # for 95% normal approx
    se = std / math.sqrt(n) if n > 0 else 0.0
    return mean, mean - z*se, mean + z*se, std

# === Main evaluation loop ===
def evaluate_models(X, y, model_names=None, n_splits=5):
    if model_names is None:
        model_names = list(MODEL_ZOO.keys())

    results = {}

    for mname in model_names:
        print("="*60)
        print(f"[EVALUATING] {mname}")
        # 1) Extract features
        feats = extract_features_for_model(mname, X)
        print(f"[INFO] Features shape for {mname}: {feats.shape}")

        # 2) 5-fold CV with LogisticRegression
        skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)
        f1s, aucs, mccs, kappas = [], [], [], []

        for train_idx, test_idx in skf.split(feats, y):
            X_train, X_test = feats[train_idx], feats[test_idx]
            y_train, y_test = y[train_idx], y[test_idx]

            scaler = StandardScaler()
            X_train_s = scaler.fit_transform(X_train)
            X_test_s = scaler.transform(X_test)

            clf = LogisticRegression(max_iter=2000, solver='lbfgs')
            clf.fit(X_train_s, y_train)

            y_pred = clf.predict(X_test_s)
            y_proba = clf.predict_proba(X_test_s)

            f1, auc, mcc, kappa = compute_fold_metrics(y_test, y_pred, y_proba)
            f1s.append(f1)
            aucs.append(auc)
            mccs.append(mcc)
            kappas.append(kappa)

        # aggregate metrics and CI (normal approx across folds)
        f1_mean, f1_ci_low, f1_ci_high, f1_std = mean_and_normal_ci(f1s)
        auc_mean, auc_ci_low, auc_ci_high, auc_std = mean_and_normal_ci(aucs)
        mcc_mean, mcc_ci_low, mcc_ci_high, mcc_std = mean_and_normal_ci(mccs)
        kappa_mean, kappa_ci_low, kappa_ci_high, kappa_std = mean_and_normal_ci(kappas)

        results[mname] = {
            "F1_scores": f1s, "F1_mean": f1_mean, "F1_ci": (f1_ci_low, f1_ci_high), "F1_std": f1_std,
            "AUC_scores": aucs, "AUC_mean": auc_mean, "AUC_ci": (auc_ci_low, auc_ci_high), "AUC_std": auc_std,
            "MCC_scores": mccs, "MCC_mean": mcc_mean, "MCC_ci": (mcc_ci_low, mcc_ci_high), "MCC_std": mcc_std,
            "Kappa_scores": kappas, "Kappa_mean": kappa_mean, "Kappa_ci": (kappa_ci_low, kappa_ci_high), "Kappa_std": kappa_std,
        }

        # Print summary
        print(f"[RESULT] {mname} -- F1 mean: {f1_mean:.4f}, 95% CI (normal approx): [{f1_ci_low:.4f}, {f1_ci_high:.4f}]")
        print(f"         AUC mean: {auc_mean:.4f}, 95% CI: [{auc_ci_low:.4f}, {auc_ci_high:.4f}]")
        print(f"         MCC mean: {mcc_mean:.4f}, 95% CI: [{mcc_ci_low:.4f}, {mcc_ci_high:.4f}]")
        print(f"         Cohen's Kappa mean: {kappa_mean:.4f}, 95% CI: [{kappa_ci_low:.4f}, {kappa_ci_high:.4f}]")
        print()

        # Visualize an internal layer feature maps for a sample image (first image)
        try:
            visualize_feature_maps_generic(mname, X[0], conv_index=3, num_maps=8)
        except Exception as e:
            print(f"[WARN] Visualization failed for {mname}: {e}")

    return results

# === RUN everything ===
if __name__ == "__main__":
    data_path = r'F:/Dataset/b_u_i_dataset/Dataset_BUSI_with_GT'  # change if needed
    print("[INFO] Loading images...")
    X, y = load_images_pil(data_path)
    if len(X) == 0:
        raise RuntimeError("No images loaded. Check data_path and folder structure.")
    X, y = shuffle(X, y, random_state=42)

    print(f"[INFO] Loaded {len(X)} images. Classes distribution:", np.bincount(y))

    # Run evaluation for all models in MODEL_ZOO
    results = evaluate_models(X, y, model_names=list(MODEL_ZOO.keys()), n_splits=5)

    # Example: produce Grad-CAM and LIME for first benign and malignant image for each model
    benign_idx = np.where(y == 0)[0][0]
    malignant_idx = np.where(y == 1)[0][0]
    benign_img = X[benign_idx].astype(np.uint8)
    malignant_img = X[malignant_idx].astype(np.uint8)

    for mname in MODEL_ZOO.keys():
        try:
            print(f"[INFO] Grad-CAM for {mname} on benign image")
            g = compute_gradcam_generic(mname, benign_img)
            plt.imshow(cv2.cvtColor(g, cv2.COLOR_BGR2RGB))
            plt.title(f"Grad-CAM {mname} (benign)")
            plt.axis('off')
            plt.show()
        except Exception as e:
            print(f"[WARN] Grad-CAM failed for {mname} benign: {e}")

        try:
            print(f"[INFO] LIME for {mname} on benign image")
            explain_with_lime_generic(mname, benign_img, num_samples=500)  # reduce samples for speed
        except Exception as e:
            print(f"[WARN] LIME failed for {mname} benign: {e}")

        try:
            print(f"[INFO] Grad-CAM for {mname} on malignant image")
            g = compute_gradcam_generic(mname, malignant_img)
            plt.imshow(cv2.cvtColor(g, cv2.COLOR_BGR2RGB))
            plt.title(f"Grad-CAM {mname} (malignant)")
            plt.axis('off')
            plt.show()
        except Exception as e:
            print(f"[WARN] Grad-CAM failed for {mname} malignant: {e}")

        try:
            print(f"[INFO] LIME for {mname} on malignant image")
            explain_with_lime_generic(mname, malignant_img, num_samples=500)
        except Exception as e:
            print(f"[WARN] LIME failed for {mname} malignant: {e}")

    print("[DONE] Evaluation complete.")
```
